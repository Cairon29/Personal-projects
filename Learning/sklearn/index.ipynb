```python
# _____________ EJERCICIO DE REGRESION LINEAL _____________

# X = características (horas de estudio, horas de sueño)
X = np.array([
    [2, 5],   # Estudiante 1
    [3, 6],   # Estudiante 2
    [4, 4],   # Estudiante 3
    [5, 7],   # Estudiante 4
])
y = np.array([60, 70, 65, 80])

kf = KFold(n_splits=2, shuffle=True, random_state=42)
model = LinearRegression()

errors = []

for train_index, test_index in kf.split(X):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
    
    model.fit(X_train, y_train)
    
    y_pred = model.predict(X_test)
    
    mse = mean_squared_error(y_test, y_pred)
    errors.append(mse)
    
    nuevas_estadisticas = confusion_matrix(y_test, y_pred)
    report = classification_report(y_train, y_test)
    print('\n\n\n matriz de confusion \n', nuevas_estadisticas)
    print('\n\n\n clasification report \n', report)
    
    print("TRAIN INDEX:", train_index, "TEST INDEX:", test_index)
    print("valores resultados training: ", y_train)
    print("Valores reales:", y_test)
    print("Predicciones :", np.round(y_pred, 2))
    print("Error (MSE):", round(mse, 2))
    print("---------------")

print("Error promedio en todos los folds:", round(np.mean(errors), 2))
```
